## What was trained

The [UCI smartphone activity dataset](https://archive.ics.uci.edu/dataset/240/human+activity+recognition+using+smartphones) has six activity classes and 561 published features derived from accelerometers and gyroscopes. Official test subjects remain outside training and validation. Validation subjects are drawn from the official training population; all splits are subject-disjoint within each seed. Standardization and PCA are fitted using selected training rows only.

The three conditions use 128 or 512 training examples, 128 validation examples and 256 test examples. Five seeds vary subject allocation, sample selection, PCA and initialization. These are overlapping views of one dataset, not independent datasets. Two timing repeats of each seed do not provide additional learning samples.

The model has four or eight independent two-qubit patches. Each patch has two independent trainable angles, each appearing twice among noncommuting fixed rotations and CZ gates. Its mean-Z feature feeds a jointly trained six-class softmax head. **Eight or sixteen total logical qubits do not form a globally entangled state: one two-qubit device could process these patches sequentially.** Increasing from eight to sixteen parameters also increases the PCA input width and classical head width, so it does not isolate parameter count alone.

GroupMemory constructs an exact local Fourier response. Qiskit receives the same factorization, 25-point support, cached diagonal blocks, exact input identities and response lifetime. Both pay for matching native off-grid checks. The direct classical adjoint computes the identical model without constructing responses. Every method receives the same patch architecture; the experiment does not assume an enormous joint response tensor across all sixteen parameters.

## Total runtime and held-out accuracy

```python

fig, axes = plt.subplots(1, 2, figsize=(14, 5.1))
positions = np.arange(len(conditions))
width = .12
for number, method in enumerate(methods):
    runtime, accuracy = [], []
    for condition in conditions:
        times = eligible_values(condition, method, "process_wall_s")
        scores = eligible_values(condition, method, "balanced_accuracy")
        runtime.append(float(np.median(times)) if len(times) else np.nan)
        accuracy.append(100*float(np.mean(scores)) if len(scores) else np.nan)
    offset = (number-(len(methods)-1)/2)*width
    axes[0].bar(positions+offset, runtime, width, color=colors[number], label=labels[method])
    axes[1].bar(positions+offset, accuracy, width, color=colors[number], label=labels[method])
axes[0].set(title="Complete fresh-process runtime", ylabel="Seconds · median of paired seed medians")
axes[1].set(title="Held-out subject classification", ylabel="Test balanced accuracy (%)")
axes[0].set_ylim(bottom=0)
axes[1].set_ylim(0, 100)
axes[1].axhline(100/6, color="#333333", linestyle="--", linewidth=1.2)
axes[1].text(-.48, 100/6+1.6, "Uniform chance: 16.7%", color="#333333", fontsize=9)
for axis in axes:
    axis.set_xticks(positions, [condition_labels[name].replace(" · ", "\n") for name in conditions])
    axis.grid(axis="y", alpha=.18)
fig.legend(*axes[0].get_legend_handles_labels(), loc="lower center", ncol=3,
           bbox_to_anchor=(.5, -.035), frameon=False)
fig.subplots_adjust(bottom=.23, wspace=.27)
fig.savefig("runtime_accuracy.png", dpi=160, bbox_inches="tight")
plt.show()
display(Markdown(
    "Runtime bars start at zero and include the entire fresh child process. "
    "The common dataset download/preparation is shown separately below. "
    "Accuracy averages each dataset seed once; two timing repeats do not double the test sample. "
    "No confidence interval is implied."
))
```

## Learning and independent parameter updates

```python

fig, axes = plt.subplots(1, max(1, len(conditions)), figsize=(4.6*max(1, len(conditions)), 3.8),
                         squeeze=False, sharey=True)
for axis, condition in zip(axes[0], conditions):
    selected = [row for row in data["learning"] if row["condition"] == condition]
    if not selected:
        axis.text(.5, .5, "No complete seed traces", ha="center", va="center", transform=axis.transAxes)
        continue
    common_steps = sorted(set.intersection(*[
        {item["step"] for item in row["training"]} for row in selected
    ]))
    train = [np.mean([
        next(item["training_regularized_log_loss"] for item in row["training"] if item["step"] == step)
        for row in selected]) for step in common_steps]
    validation_steps = sorted(set.intersection(*[
        {item["step"] for item in row["validation"]} for row in selected
    ]))
    validation = [np.mean([
        next(item["log_loss"] for item in row["validation"] if item["step"] == step)
        for row in selected]) for step in validation_steps]
    axis.plot(np.array(common_steps)-1, train, color=colors[0], linewidth=2, label="Training + L2")
    axis.plot(validation_steps, validation, color=colors[1], linewidth=2, label="Validation")
    axis.axhline(np.log(6), color="#777777", linewidth=1, linestyle=":", label="Uniform log loss")
    axis.set(title=condition_labels[condition].replace(" · ", "\n"),
             xlabel="Completed Adam updates", ylim=(0, None))
    axis.grid(alpha=.15)
axes[0, 0].set_ylabel("Mean cross-entropy across dataset seeds")
axes[0, 0].legend(frameon=False, fontsize=9)
fig.tight_layout()
fig.savefig("learning.png", dpi=160, bbox_inches="tight")
plt.show()
display(Markdown(
    "Only one timing repeat per seed contributes to these curves. They show "
    "GroupMemory; the matched Qiskit and adjoint trajectories are checked separately. "
    "Training loss includes the small head L2 penalty and is measured before each "
    "update; validation loss is unregularized and measured after the indicated number "
    "of updates. Each run's lowest validation loss selects its checkpoint before "
    "one final test evaluation. Sixty updates are a bounded learning demonstration, "
    "not a convergence guarantee."
))
if audit:
    table([{
        "Condition": condition_labels[condition],
        "Seeds": item["observed_seeds"],
        "Angles moved by final update": f'{item["min_moved_parameters"]}–{item["max_moved_parameters"]}',
        "Angles moved at selected checkpoint": f'{item["min_selected_moved_parameters"]}–{item["max_selected_moved_parameters"]}',
        "Selected updates": ", ".join(map(str, item["selected_steps"])),
        "Median angle displacement norm": item["median_displacement_norm"],
    } for condition, item in audit["parameter_motion"].items()],
    {"Median angle displacement norm": "{:.4f}".format})
```

## Classical model selection

```python

rows = []
for condition in conditions:
    for method in ("classical_logistic", "classical_rbf", "classical_mlp"):
        selected = [item for item in data["classical_views"]
                    if item["condition"] == condition and item["method"] == method]
        if not selected:
            continue
        counts = Counter(item["selected_view"] for item in selected)
        losses = {
            view: [candidate["validation_log_loss"] for item in selected
                   for candidate in item["candidates"] if candidate["view"] == view]
            for view in ("angles", "full561")
        }
        warnings = sum(len(candidate["warnings"]) for item in selected for candidate in item["candidates"])
        rows.append({
            "Condition": condition_labels[condition], "Classical family": labels[method],
            "Angles selected": counts["angles"], "All 561 selected": counts["full561"],
            "Angles validation loss": float(np.mean(losses["angles"])),
            "All 561 validation loss": float(np.mean(losses["full561"])),
            "Recorded fit warnings": warnings,
        })
table(rows, {"Angles validation loss": "{:.4f}".format,
             "All 561 validation loss": "{:.4f}".format})
display(Markdown(
    "Each classical family fits both feature views and pays for both fits and "
    "validation predictions. The view with lower validation log loss is selected. "
    "Hyperparameters are fixed in advance; the test set selects neither the view "
    "nor the learner. The table counts each dataset seed once. "
    "The MLP angle-view fits reached the 200-iteration limit and emitted "
    "convergence warnings, including some selected fits. These are bounded "
    "classical fits, not certified converged models; the raw records preserve "
    "each warning. Classical controls can select all 561 features, so their "
    "task comparison also involves a different input representation."
))
```

## Complete timing scope

Every method starts in a fresh process with an empty cache. The main process clock includes interpreter startup, imports, dataset reading, train-only preprocessing, model construction, compilation, response calibration, correctness checks, bookkeeping, training, validation selection, final inference, result writing and shutdown. The response methods precompute validation responses when the update count reaches their 25-point grid size; that preparation is charged. Test responses are not warmed.

Task-specific classical methods pay for fitting both the common angle-feature view and all 561 standardized features. Common initial download/text preparation and campaign orchestration are reported separately.

```python

setup = data["provenance"].get("common_setup_timings", {})
common_s = sum(value for value in setup.values() if isinstance(value, (int, float)))
table([{"Common acquisition component": key, "Seconds": value} for key, value in setup.items()],
      {"Seconds": "{:.6f}".format})
display(Markdown(
    f'The recorded one-time download, extraction, text parsing and NPZ writing components sum to '
    f'**{common_s:.6f} seconds**. This is a component sum, not a full acquisition wall clock. '
    'Manual coordination and small uninstrumented acquisition tasks are outside that sum. '
    'This common setup is paid once, outside every method '
    'clock. Each timed process still reads the NPZ and fits its own training-only '
    'scaler/PCA. The original raw-sensor feature-extraction time is unavailable.'
))
cold = []
for condition in conditions:
    for method in methods:
        times = eligible_values(condition, method, "process_wall_s")
        if len(times):
            cold.append({"Condition": condition_labels[condition], "Method": labels[method],
                         "Prepared-data process s": float(np.median(times)),
                         "Cold start s, common setup added once": float(np.median(times))+common_s})
table(cold, {"Prepared-data process s": "{:.3f}".format,
             "Cold start s, common setup added once": "{:.3f}".format})
display(Markdown(
    "The cold-start column is an accounting sum using the observed common acquisition "
    "cost; it is not a separately repeated download benchmark. A shared additive cost "
    "reduces speedup ratios but does not change the measured difference in seconds."
))
wallrows = [{"Scope": "Sum of all saved attempt process clocks",
             "Seconds": data["attempt_process_wall_sum_s"]}]
for name, receipt in data["operational_wall_receipts"].items():
    seconds = receipt.get("seconds", receipt.get("wall_s"))
    if isinstance(seconds, (int, float)):
        wallrows.append({"Scope": name, "Seconds": seconds})
table(wallrows, {"Seconds": "{:.3f}".format})
display(Markdown(
    "Campaign and preflight receipts are separately reported operational costs. "
    "The campaign clock includes its child attempts, so these overlapping clocks "
    "must not be summed. Package installation, implementation and original dataset "
    "feature extraction are outside the method benchmark."
))
```

## Coverage, correctness and reproducibility

```python

table([{
    "Condition": condition_labels[condition],
    **{state: data["coverage"][condition].get(state, 0)
       for state in ("SUCCESS", "ERROR", "TIMEOUT", "UNATTEMPTED")},
} for condition in conditions])
if data["failures"]:
    table([{"Condition": row["condition"], "Seed": row["seed"], "Repeat": row["repeat"],
            "Method": labels[row["method"]], "Status": row["status"], "Error": row["error"]}
           for row in data["failures"]])
else:
    display(Markdown("No missing, failed or timed-out attempts appear in this saved plan."))
table([{"Package": key, "Version": value}
       for key, value in data["environment"]["packages"].items()])
display(Markdown(
    f'Platform: `{data["environment"]["platform"]}`. '
    f'Each fresh worker used **{data["environment"]["threads_per_worker"]} computational thread** '
    'and attempts ran serially. This is ideal CPU simulation; no QPU, hardware noise, '
    'finite-shot measurement, queue time or hardware speed advantage was measured.'
))
display(Markdown(
    f'This notebook embeds a saved-data snapshot with SHA256 '
    f'`{SNAPSHOT_SHA256}`. It contains hashes for **{len(data["attempt_hashes"])}** '
    f'attempt files and **{len(data["source_hashes"])}** source files. '
    'The public results directory contains the full receipts and audit; rerunning '
    'this notebook regenerates displays only.'
))
```

The strict same-model comparison matches saved optimizer trajectories, gradients, checkpoint selection and final predictions to **1e−10**. It does not save every intermediate feature Jacobian. Independent unit tests also compare patch outputs, full complex operators and derivatives with separately constructed circuits.

The comparisons describe a bounded CPU systems result on this factorized model. They do not demonstrate quantum hardware advantage, a globally entangled sixteen-qubit learner, generalization across datasets, or convergence after sixty updates. There is no frozen-quantum-feature ablation isolating the benefit of adapting quantum angles from training the classical head. All negative, incomplete and neutral comparisons remain in the report.
